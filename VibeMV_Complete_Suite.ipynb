## 💡 Tips

### For Best Results

1. **Start small:** Test with 5-10 scenes first
2. **Character consistency:** Upload reference image in Phase 1
3. **Prompt quality:** Better prompts = better results
4. **Audio sync:** Add audio in final step

### Troubleshooting

**Out of VRAM:**
- Use lower resolution
- Reduce num_frames
- Enable more optimizations

**Low quality:**
- Increase inference_steps
- Use negative prompts
- Try different method

**Too slow:**
- Use SDXL or Hybrid method
- Reduce scene count
- Lower resolution
